3. Data Transformation
   Perform the required transformations, such as changing data types, dropping unnecessary columns, or adding new columns.


```python
# dropping unnecessary column
df=df.drop('Unnamed: 0',axis=1)
print(df.head())
```

```python
# changing datatype
df['time_signature']=df['time_signature'].astype(int)
df['time_signature'].head()
```

```python
# adding new column
df['duration_seconds']=df['duration_ms']/1000
print(df.head())
```

4. Handling Null Values
   Identify null values in the dataset.
   Handle null values using mean imputation wherever applicable.

```python
# checking null values
df.isnull().sum()
```

```python
# filling null values using mean
df['acousticness']=df['acousticness'].fillna(df['acousticness'].mean())

df['danceability']=df['danceability'].fillna(df['danceability'].mean())

df['energy']=df['energy'].fillna(df['energy'].mean())

df['loudness']=df['loudness'].fillna(df['loudness'].mean())

df['tempo']=df['tempo'].fillna(df['tempo'].mean())

df['valence']=df['valence'].fillna(df['valence'].mean())
```

```python
# checking null values after filling

df.isnull().sum()
```

```python
# using apply and lambda

df['double_energy']=df['energy'].apply(lambda x:x*2)
print(df[['energy','double_energy']].head())
```

5. Insight Generation
   Generate meaningful insights from the dataset using the concepts taught in class.
   Examples include sales trends over time and region-wise revenue analysis, depending on the selected dataset.


```python
# grouping artist and finding average danceability
artist_danceability=df.groupby('artist')['danceability'].mean()
print(artist_danceability)
```

```python
print(artist_danceability.head(10))
```

```python
# grouping artist and finding average energy

artist_energy=df.groupby('artist')['energy'].mean()
print(artist_energy.head(10))
```

```python
# counting songs for each artist

artist_songs=df.groupby('artist')['song_title'].count()
print(artist_songs)
```

```python
# sorting number of songs

artist_songs=artist_songs.sort_values(ascending=False)
print(artist_songs.head(10))
```

```python
# grouping using target

target_analysis=df.groupby('target')[['danceability','energy','valence','tempo']].mean()
print(target_analysis)
```

```python
# sorting songs according to energy

high_energy=df.sort_values('energy',ascending=False)
print(high_energy[['song_title','artist','energy']].head(10))
```

```python
# sorting songs according to danceability

high_danceability=df.sort_values('danceability',ascending=False)
print(high_danceability[['song_title','artist','danceability']].head(10))
```

6. Data Visualization
   Visualize the generated insights using appropriate charts.


```python
# getting top 10 artists

top_artists=artist_songs.head(10)
print(top_artists)
```

```python
# bar graph

top_artists.plot(kind='bar')
plt.xlabel('Artist')
plt.ylabel('Number of Songs')
plt.title('Top 10 Artists by Number of Songs')
plt.show()
```

```python
# scatter plot

plt.scatter(df['energy'],df['danceability'])

plt.xlabel('Energy')
plt.ylabel('Danceability')
plt.title('Energy vs Danceability')
plt.show()
```

```python
# grouping target and finding average energy

target_energy=df.groupby('target')['energy'].mean()
print(target_energy)
```

```python
# 5. Distribution of Energy
# histogram

plt.hist(df['energy'],bins=10)
plt.xlabel('Energy')
plt.ylabel('Number of Songs')
plt.title('Distribution of Song Energy')
plt.show()
```
